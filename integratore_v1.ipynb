## Integratore

L'integratore RC è un semplice circuito, composto da una resistenza e un condensatore con tempo caratteristico $\tau = RC$, che, per un segnale (d.d.p.) in ingresso periodico di periodo $T$ con $T << \tau$, produce in uscita un segnale che approssima l'integrale nel tempo dell'ingresso.

L'effetto dell'integratore è particolarmente semplice da visualizzare quando la forma d'onda in ingresso è quadra: all'uscita si ha una forma d'onda che tende alla triangolare. La configurazione circuitale è del tutto simile a quella dell'esperimento sulla carica/scarica del condensatore, per la cui visualizzazione "completa" (cioè comprensiva degli asintoti di carica e scarica) occorre però $T > \tau$.

Lo script qui di seguito mostra come, al variare del rapporto $T/\tau$, che può essere modificato agendo sullo slider, il segnale in uscita **a regime** (in condizioni stazionarie), rappresentato in rosso, cambi la sua forma da quella tipica di un processo di carica/scarica alla triangolare al diminuire del rapporto $T/\tau$. La modifica della forma è accompagnata da una sensibile riduzione dell'ampiezza.

Nel grafico il tempo è misurato in unità di $\tau$ e inizialmente il periodo della forma d'onda quadra è tale che $T/\tau = 10$, in modo da visualizzare gli asintoti di carica e scarica. Per riferimento si mostrano anche:
- in verdino il segnale della carica/scarica per $T/\tau = 10$;
- in celestino la forma d'onda quadra in ingresso.

L'aspetto più interessante dello script, che è volutamente un po' nascosto, è il metodo adottato per ricostruire le forme d'onda ottenute alle varie frequenze, che sono simulate con il metodo della serie di Fourier e della funzione di trasferimento per un filtro passa-basso RC. Guardate e cercate di capire: può essere utile per il vostro assegnamento obbligatorio!


```python
from ipywidgets import interact
import ipywidgets as widgets
import numpy as np
import pylab as py

nofpoints = 1000 # numero di punti dell'array
niter = 100 # numero di iterazioni della simulazione
omegat = 10/(2*np.pi) # frequenza angolare di taglio del filtro passa-basso (corrisponde a tau = T_iniziale/10)

tarray = np.linspace(0,10,nofpoints) # crea l'array dei tempi

wave0 = np.array(nofpoints) # crea l'array della carica/scarica
wave0 = 0
omega0 = 2*np.pi/10
for i in range (1,niter,2): # itera per creare la somma di Fourier nelle condizioni T/tau = 10
        omegak = i*omega0
        guad = 1/np.sqrt(1+(omegak/omegat)**2)
        sfas = np.atan(-omegak/omegat)
        wave0 = wave0 + guad*(4/(i*np.pi)) * np.sin(omegak*tarray+sfas)

def plotta(tsutau):
    omega0 = 2*np.pi/tsutau # stabilisce quanto vale la frequenza angolare a partire dal periodo
    wave = np.array(nofpoints) # crea l'array della forma d'onda
    wave = 0
    for i in range (1,niter,2): # itera per creare la somma di Fourier
        omegak = i*omega0
        guad = 1/np.sqrt(1+(omegak/omegat)**2)
        sfas = np.atan(-omegak/omegat)
        wave = wave + guad*(4/(i*np.pi)) * np.sin(omegak*tarray+sfas)
    
    quad = np.sign(np.sin(omega0*tarray)) # crea l'array dell'onda quadra
    py.figure(figsize=(10,4))
    py.plot(tarray,quad,color='lightsteelblue') # disegna quello che deve
    py.plot(tarray,wave0,color='limegreen')
    py.plot(tarray,wave,color='red')
    py.minorticks_on()
    py.xlabel('Time  [$\\tau$]',fontsize='14')
    py.ylabel('Signal  [arb.un.]',fontsize='14')

# permette l'inteerazione attraverso lo slider
interact(
    plotta,
    tsutau = widgets.FloatSlider(value=10, min=0.1, max=10, step=0.01,description='$T/\\tau$',readout_format='.2f')
    )   
```
