# Campionamento e alias

Questo notebook mostra gli effetti del campionamento sull'acquisizione di un ipotetico segnale di forma sinusoidale, ampiezza unitaria e frequenza data. 

Il campionamento produce un record composto da $N_{pnts}$ punti. Inoltre si suppone di poter variare il sampling rate del campionatore in modo da acquisire un numero $N_T$ di periodi.

Gli effetti del campionamento discreto, seppur banalissimi e di facile comprensione, sono spesso trascurati, per cui conviene spendere un minimo di tempo per verificarli visivamente. Quello che potrete facilmente verificare giocando con gli sliders è che *non sempre* il grafico rappresenta l'andamento sinusoidale atteso e che, in determinate condizioni, il segnale assume una forma parecchio diversa dalle aspettative, a cui si dà spesso il nome di **alias**.

Osservate che, benché i concetti di campionamento siano propri delle misure digitali, gli effetti si ritrovano ogni volta che ci si trova a rappresentare una funzione idealmente continua in forma discreta. Questo si verifica **sempre** quando si usa un qualsiasi software di rappresentazione grafica e anche quando si usano display o fogli di carta con *risoluzione finita*!


```python
from ipywidgets import interact
import ipywidgets as widgets
import pylab as py
import numpy as np

# definisce la funzione che viene rappresentata in modo interattivo
def plotta(nofperiods,nofpoints):
    tarray = np.linspace(0, nofperiods,int(nofpoints)) # questo è l'array ancillare che rappresenta i tempi
    seno = np.sin(tarray*2*np.pi) # questa è la funzione
    py.plot(tarray,seno) # questo è il solito comando per il grafico
    py.xlabel('Time  [T]') # e questi sono i nomi degli assi
    py.ylabel('Signal  [arb.un.]')

# qui c'è l'operazione, specifica del pacchetto ipywidgets, che crea gli slider e permette il grafico interattivo
interact(
    plotta,
    nofperiods = widgets.FloatSlider(value=4, min=0.1, max=100, step=0.5,description='$N_T$',readout_format='.1f'),
    nofpoints = widgets.FloatSlider(value=500, min=10, max=1000, step=10, description='$N_{pnts}$',readout_format='.0f'),
    )           
```
