# Sfasamento e modalità XY dell'oscilloscopio

La modalità **XY** (o YX) dell'oscilloscopio è utile per verificare la presenza di uno sfasamento $\Delta \Phi$ tra due segnali sinusoidali, $V1(t)$ e $V2(t)$, alla stessa frequenza.

Infatti, se supponete uno sfasamento $\Delta \Phi = \pi/2$ [rad] e pensate alle componenti di un moto circolare uniforme, cioè a velocità angolare costante $\omega$, potete facilmente capire come il grafico parametrico prodotto nella modalità XY corrisponda a una orbita circolare che si richiude su se stessa. 

Il carattere circolare dell'orbita dipende in realtà dalla circostanza che il rapporto $V2/V1$ tra le ampiezze dei due segnali sia unitario e che la sensibilità dei due canali dell'oscilloscopio che si stanno visualizzando sia la stessa. Se questo non si verifica, allora si ottiene una orbita **ellittica**, con i *semiassi orientati lungo le direzioni cartesiane* nel caso $\Delta \Phi = \pi/2$ [rad].

Per sfasamenti $\Delta \Phi \neq \pi/2$ [rad] e $\Delta \Phi \neq 0$, la figura ottenuta è ancora ellittica, ma i suoi semiassi non sono più orientati lungo le direzioni cartesiane. La dipendenza dell'angolo formato da un semiasse con una direzione cartesiana con lo sfasamento è molto blanda e non lineare, per cui l'inclinazione dell'ellisse fornisce con difficoltà un'informazione sullo sfasamento $\Delta \Phi$.

L'ellisse degenera in un segmento *inclinato* per $\Delta \Phi = 0$ (ovvero $\Delta \Phi = \pi$ [rad]). Poiché la degenerazione è facile da osservare, l'utilità principale delle osservazioni in modalità XY è l'individuazione delle *condizioni per cui lo sfasamento è nullo*. Questo è quello di cui ci serviremo in genere a Lab2.

Tuttavia è ancora possibile dedurre il valore di $\Delta \Phi$ dalle caratteristiche dell'ellisse visualizzata. Si possono per esempio individuare due valori, $y_1$ e $y_2$, corrispondenti rispettivamente al *massimo del segnale in direzione verticale*, $V2_{max}$, e al valore di $V2$ all'istante per cui $V1 = V_{max}$, tali che
$$ \Delta \Phi = \textrm{arccos}(y_2/y_1)\;.$$.

Lo script che segue mostra l'ellisse che si forma quando si osservano in modalità XY i due segnali
$$V1(t) = V1 \cos(\omega t)$$
$$V2(t)=V2 \cos(\omega t + \Delta \Phi)\;.$$.

Il valore di $\Delta \Phi$ può essere variato agendo su uno slider. Nel grafico compaiono anche:
- l'indicazione di un semiasse dell'ellisse e di un asse, tramite linee dotted rosse;
- l'indicazione dei punti e delle coordinate necessarie per individuare $y_1$ e $y_2$ (linee dashed verdi).

Lo script determina numericamente (vedete come e tenete presente che non funziona per un rapporto unitario tra le ampiezze) l'angolo tra il semiasse indicato in figura e l'asse cartesiano orizzontale e lo sfasamento determinato dal rapporto $y_1/y_2$.

Se volete, date un'occhiata a come sono eseguiti i calcoli e a tutto il resto!




```python
from ipywidgets import interact
import ipywidgets as widgets
import pylab as py
import numpy as np

nofperiods = 4 # numero di periodi considerati
nofpoints = 1000 # numero di punti degli arrays
amplratio = 0.8 # rapporto tra le ampiezze V2/V1 (diverso da uno può favorire la leggibilità, uguale a uno crea problemi nelle soluzioni numeriche

xred=np.array([0.,0.]);yred=np.array([0.,0.]) # crea gli array di coordinate necessarie per disegnare la linea dotted rossa orizzontale
horred=np.array([0.,0.]);vertred=np.array([0.,0.]) # anche qui


# definisce la funzione che viene rappresentata in modo interattivo
def plotta(deltaphi):
    tarray = np.linspace(0, nofperiods,int(nofpoints)) # questo è l'array ancillare che rappresenta i tempi
    V1 = np.sin(tarray*2*np.pi) # questa è la funzione V1
    V2 = amplratio*np.sin(tarray*2*np.pi+deltaphi*np.pi) # questa è la funzione V2, sfasata di deltaphi rispetto a V1
    py.plot(V1,V2) # disegna il grafico parametrico V2 vs V1
    py.xlabel('V1  [arb.un.]') # e questi sono i nomi degli assi
    py.ylabel('V2  [arb.un.]')
    py.xlim(-1.05,1.05) # qui fissa le scale
    py.ylim(-1.05,1.05)
    py.grid() # qui disegna la griglia che somiglia alla graticola dell'oscilloscopio

    squared=np.sqrt(V1**2+V2**2) # crea un array che contiene la distanza dei punti dell'ellisse dall'origine
    maxpoint=np.argmax(squared) # trova la coordinata del massimo della distanza, che dovrebbe essere l'afelio rispetto al centro
    if (deltaphi < 0.51): # questi if else sono delle orrende pezze a un problema di precisione matematica finita...
        xred[1]=np.abs(V1[maxpoint])
    else: 
        xred[1]=-np.abs(V1[maxpoint])
    if (deltaphi > -0.5):
        yred[1]=np.abs(V2[maxpoint])
    else:
        yred[1]=-np.abs(V2[maxpoint])
    horred[1]=xred[1]
    py.plot(horred,vertred,color='red',linestyle='dotted') # disegna un mezzo asse orizzontale con linea dotted rossa
    py.plot(xred,yred,color='red',linestyle='dotted') # disegna mezzo semiasse con linea dotted rossa
    angolo=np.atan2(yred[1],xred[1])*180/np.pi # calcola l'angolo in gradi tra semiasse e asse orizzontale
    print('L\'angolo tra semiasse e orizzontale (tra i segmenti dotted in rosso), calcolato numericamente, è '+f'{angolo:.0f}'+' gradi')
    vlinex=np.array([np.max(V1),np.max(V1)]) # trova le coordinate per disegnare la linea verticale dashed verde
    vliney=np.array([-1.05,V2[np.argmax(V1)]]) # anche qui
    py.plot(vlinex,vliney,color='green',linestyle='dashed') # e la disegna
    hlinex=np.array([-1.05,V1[np.argmax(V2)]]) # stessa cosa per la linea orizzontale dashed verde
    hliney=np.array([np.max(V2),np.max(V2)])
    py.plot(hlinex,hliney,color='green',linestyle='dashed')
    hhlinex=np.array([-1.05,np.max(V1)]) # e per l'altra linea orizzontale dashed verde
    hhliney=np.array([V2[np.argmax(V1)],V2[np.argmax(V1)]])
    py.plot(hhlinex,hhliney,color='green',linestyle='dashed')
    py.text(-1,np.max(V2),'$y_2 = V2_{max}$',fontsize='14') # scrive sulla posizione della coordinata y_2
    py.text(-1,V2[np.argmax(V1)],'$y_1$',fontsize='14') # scrive sulla posizione della coordinata orizzontale
    py.text(np.max(V1),-1, '$V1_{max}$',fontsize='14') # scrive sulla posizione orizzontale del massimo di V1
    sfasamentonumerico=np.acos(V2[np.argmax(V1)]/np.max(V2)) # calcola numericamente lo sfasamento in radianti
    print('Lo sfasamento calcolato numericamente è '+f'{sfasamentonumerico/np.pi:.2f}'+' [pi rad]'+' (corrispondente a '+f'{sfasamentonumerico*180/np.pi: .1f}'+' [gradi])')
    
# qui c'è l'operazione, specifica del pacchetto ipywidgets, che crea lo slider e permette il grafico interattivo
interact(
    plotta,
    deltaphi = widgets.FloatSlider(value=0, min=-1, max=1, step=0.01,description='$\\Delta \\Phi$  [$\\pi$ rad]',readout_format='.2f')
    )   
```
