## Voltmetro Reale

Data una data intensità di corrente fissata $I$ che scorre in un ramo resistivo di resistenza $R$, la d.d.p. $V$ ai capi del ramo è:

$V = R I$.

Quando la d.d.p. viene letta da un voltmetro **reale**, la resistenza $R$ viene a trovarsi in parallelo con la resistenza interna del voltmetro $r_V$.
Di conseguenza, la lettura della d.d.p. viene perturbata e il suo valore risulta

$V_{pert} = R_{eq,//} I$, 

con

$R_{eq,//} = \frac{R r_V}{R+r_V}$.

Nel notebook, dati i valori di $R$, $I$, e $r_V$ (scelta da un menu a tendina con opzioni *tipiche*), vengono calcolati a scopo di confronto i valori di $V$ e $V_{pert}$.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Creazione di un menu a tendina con valori prefissati della resistenza interna del voltmetro
menu = widgets.Dropdown(
    options=[('10 Mohm',1e7), ('200 kohm',2e5), ('1 Mohm', 1e6), ('50 ohm',50)],
    value=1e7, # Valore iniziale predefinito
    description='$r_V$',
    disabled=False,
)

# Creazione delle caselle di testo per l'input di intensità di corrente e resistenza del ramo
casella_input_1 = widgets.Text(
    value='',
    placeholder='Resistenza del ramo (in ohm):',
    description='$R$ (in ohm):',
    disabled=False
)

casella_input_2 = widgets.Text(
    value='',
    placeholder='Intensità di corrente nel ramo (in mA):',
    description='$I$ (in mA):',
    disabled=False
)

# Creazione del pulsante di invio
pulsante = widgets.Button(
    description='Calcola',
    button_style='success', # Verde
    tooltip='Clicca per confermare',
    icon='check'
)

# Definizione area di output per mostrare il risultato
area_output = widgets.Output(layout={'border': '1px solid red'})

# Funzione che gestisce l'evento del click sul pulsante (fa anche i calcolini)
def click(b): 
    with area_output: # Questa istruzione è specifica per ipywidgets
        clear_output() # Pulisce i messaggi precedenti
        R = float(casella_input_1.value) # Prende il valore di R
        I = float(casella_input_2.value) # quello di I
        r_V=float(menu.value) # quello di r_V
        V_imperturbato=R*I*1e-3 # e fa i calcolini
        V_perturbato=((R*r_V)/(R+r_V))*I*1e-3        
        print(f'Se il voltmetro non perturbasse il circuito, la lettura sarebbe: {V_imperturbato:.2e} [V]')
        print(f'Invece con il voltmetro in uso la lettura è: {V_perturbato:.2e} [V]')

# Collegamento della funzione al pulsante
pulsante.on_click(click)
    
# Stampa sul notebook dei vari elementi grafici
display(menu,casella_input_1, casella_input_2,pulsante,area_output)
```

## Amperometro reale

Data una data d.d.p. fissata $\Delta V$ ai capi di un ramo resistivo di resistenza $R$, l'intensità di corrente $I$ che scorre nel ramo è:

$I = \frac{\Delta V}{R}$.

Quando la corrente viene letta da un amperometro **reale**, la resistenza $R$ viene a trovarsi in serie con la resistenza interna dell'amperometro, $r_A$.
Di conseguenza, la lettura dell'intensità di corrente viene perturbata e il suo valore risulta

$I_{pert} = \frac{\Delta V}{R+r_A}$. 

Nel notebook, dati i valori di $R$, $\Delta V$, e $r_A$, vengono calcolati a scopo di confronto i valori di $I$ e $I_{pert}$.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Creazione delle caselle di testo per l'input di d.d.p, di resistenza del ramo e resistenza interna
casella_input_0 = widgets.Text(
    value='',
    placeholder='Resistenza interna dello amperometro (in ohm):',
    description='$r_A$ (in ohm):',
    disabled=False
)

casella_input_1 = widgets.Text(
    value='',
    placeholder='Resistenza del ramo (in ohm):',
    description='$R$ (in ohm):',
    disabled=False
)

casella_input_2 = widgets.Text(
    value='',
    placeholder='d.d.p. ai capi del ramo (in V):',
    description='Delta $V$ (in V):',
    disabled=False
)

# Creazione del pulsante di invio
pulsante = widgets.Button(
    description='Calcola',
    button_style='success', # Verde
    tooltip='Clicca per confermare',
    icon='check'
)

# Definizione area di output per mostrare il risultato
area_output = widgets.Output(layout={'border': '1px solid red'})

# Funzione che gestisce l'evento del click sul pulsante (fa anche i calcolini)
def click(b): 
    with area_output: # Questa istruzione è specifica per ipywidgets
        clear_output() # Pulisce i messaggi precedenti
        R = float(casella_input_1.value) # Prende il valore di R
        DV = float(casella_input_2.value) # quello di Delta V
        r_A=float(casella_input_0.value) # quello di r_A
        I_imperturbato=DV/R # e fa i calcolini
        I_perturbato=DV/(R+r_A)        
        print(f'Se amperometro non perturbasse il circuito, la lettura sarebbe: {I_imperturbato:.2e} [A]')
        print(f'Invece con lo amperometro in uso la lettura è: {I_perturbato:.2e} [A]')

# Collegamento della funzione al pulsante
pulsante.on_click(click)
    
# Stampa sul notebook dei vari elementi grafici
display(casella_input_0,casella_input_1, casella_input_2,pulsante,area_output)
```

## Esperimento esercitazione 1

Nell'esercitazione 1 eravate chiamati a verificare gli effetti della resistenza interna del voltmetro reale costruendo un semplice circuitino che comprendeva un generatore di d.d.p. $\Delta V$ (supposto **ideale** in questo contesto), un amperometro **reale** che misurava l'intensità di corrente $I$, una resistenza $R$, e un voltmetro **reale** che misurava la d.d.p. $V$ ai capi della resistenza. 

In particolare, dovevate verificare se la lettura del voltmetro reale, $V$ era o meno equivalente al prodotto $R I$.

La situazione concettuale è la stessa esaminata prima, ma stavolta l'intensità di corrente erogata dal generatore e letta dall'amperometro, $I$, **dipende** da $R$ e da $r_V$. Infatti è

$I =\frac{\Delta V}{R_{eq,//}}$

con

$R_{eq,//} = \frac{R r_V}{R+r_V}$.

Dunque il prodotto $R I$ è

$R I = \frac{R}{R_{eq,//}}\Delta V = \frac{R+r_V}{r_V} \Delta V$.

Essendosi formato un partitore di corrente, solo una frazione $I_R$ dell'intensità di corrente $I$ fluisce nella resistenza:

$I_R = \frac{r_V}{R+r_V} I$.

Quindi la lettura del voltmetro è

$V = R I_R = \Delta V$,

come ovvio.

Il codice qui sotto esegue i calcoli e mostra il confronto tra il prodotto $R I$ e la lettura $V$.

**Qui non viene considerata la presenza dell'amperometro reale**, che sarà incluso nel prossimo punto.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Creazione di un menu a tendina con valori prefissati della resistenza interna del voltmetro
menu = widgets.Dropdown(
    options=[('10 Mohm',1e7), ('200 kohm',2e5), ('1 Mohm', 1e6), ('50 ohm',50)],
    value=1e7, # Valore iniziale predefinito
    description='$r_V$',
    disabled=False,
)

# Creazione delle caselle di testo per l'input di d.d.p. erogata dal generatore (ideale) e resistenza del ramo
casella_input_1 = widgets.Text(
    value='',
    placeholder='Resistenza del ramo (in ohm):',
    description='$R$ (in ohm):',
    disabled=False
)

casella_input_2 = widgets.Text(
    value='',
    placeholder='d.d.p. erogata dal generatore (in V):',
    description='Delta$V$ (in V):',
    disabled=False
)

# Creazione del pulsante di invio
pulsante = widgets.Button(
    description='Calcola',
    button_style='success', # Verde
    tooltip='Clicca per confermare',
    icon='check'
)

# Definizione area di output per mostrare il risultato
area_output = widgets.Output(layout={'border': '1px solid red'})

# Funzione che gestisce l'evento del click sul pulsante (fa anche i calcolini)
def click(b): 
    with area_output: # Questa istruzione è specifica per ipywidgets
        clear_output() # Pulisce i messaggi precedenti
        R = float(casella_input_1.value) # Prende il valore di R
        DV = float(casella_input_2.value) # quello di Delta V
        r_V=float(menu.value) # quello di r_V
        Req=R*r_V/(R+r_V) # calcola R_eq
        I = DV/Req # calcola I
        erreI= R*I # calcola R I
        print(f'Con il voltmetro in uso, la intensità di corrente letta dallo amperometro è: {I*1e3:.2e} [mA]')
        print(f'Il prodotto RI è: {erreI:.2e} [V]')
        print(f'La lettura del voltmetro è: {DV:.2e} [V]')

# Collegamento della funzione al pulsante
pulsante.on_click(click)
    
# Stampa sul notebook dei vari elementi grafici
display(menu,casella_input_1, casella_input_2,pulsante,area_output)
```

## Esperimento considerando anche l'amperometro reale

La presenza dell'amperometro **reale** con resistenza interna $r_A$ modifica (di poco, eh!) i risultati per due motivi:

- l'intensità di corrente erogata dal generatore diventa

$I =\frac{\Delta V}{R_{eq,//}+r_A}$;

- la d.d.p. ai capi della resistenza, e quindi letta dal voltmetro, diventa

$V = \Delta V - r_A I = \Delta V(1- \frac{r_A}{R_{eq,//}+r_A}) = \Delta V \frac{R_{eq,//}}{R_{eq,//}+r_A}$.

Il codice qui sotto esegue i calcoli e mostra il confronto tra il prodotto $R I$ e la lettura $V$ in questo caso (scegliete valori ragionevoli per $r_A$!). Vedrete che è abbastanza difficile ottenere risultati diversi dal caso precedente...

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Creazione di un menu a tendina con valori prefissati della resistenza interna del voltmetro
menu = widgets.Dropdown(
    options=[('10 Mohm',1e7), ('200 kohm',2e5), ('1 Mohm', 1e6), ('50 ohm',50)],
    value=1e7, # Valore iniziale predefinito
    description='$r_V$',
    disabled=False,
)

# Creazione delle caselle di testo per l'input di resistenza interna dell'amperometro, d.d.p. erogata dal generatore (ideale) e resistenza del ramo
casella_input_0 = widgets.Text(
    value='',
    placeholder='Resistenza interna dello amperometro (in ohm):',
    description='$r_A$ (in ohm):',
    disabled=False
)

casella_input_1 = widgets.Text(
    value='',
    placeholder='Resistenza del ramo (in ohm):',
    description='$R$ (in ohm):',
    disabled=False
)

casella_input_2 = widgets.Text(
    value='',
    placeholder='d.d.p. erogata dal generatore (in V):',
    description='Delta$V$ (in V):',
    disabled=False
)

# Creazione del pulsante di invio
pulsante = widgets.Button(
    description='Calcola',
    button_style='success', # Verde
    tooltip='Clicca per confermare',
    icon='check'
)

# Definizione area di output per mostrare il risultato
area_output = widgets.Output(layout={'border': '1px solid red'})

# Funzione che gestisce l'evento del click sul pulsante (fa anche i calcolini)
def click(b): 
    with area_output: # Questa istruzione è specifica per ipywidgets
        clear_output() # Pulisce i messaggi precedenti
        R = float(casella_input_1.value) # Prende il valore di R
        DV = float(casella_input_2.value) # quello di Delta V
        r_V=float(menu.value) # quello di r_V
        r_A=float(casella_input_0.value) # quello di r_A
        Req=R*r_V/(R+r_V) # calcola R_eq
        I = DV/(Req+r_A) # calcola I
        erreI= R*I # calcola R I
        V = DV*Req/(Req+r_A) # calcola V letta dal voltmetro       
        print(f'Con il voltmetro e amperometro in uso, la intensità di corrente letta dallo amperometro è: {I*1e3:.2e} [mA]')
        print(f'Il prodotto RI è: {erreI:.2e} [V]')
        print(f'La lettura del voltmetro è: {V:.2e} [V]')

# Collegamento della funzione al pulsante
pulsante.on_click(click)
    
# Stampa sul notebook dei vari elementi grafici
display(menu, casella_input_0, casella_input_1, casella_input_2,pulsante,area_output)
```
